```python
import os 
```

```python
os.environ["OPENAI_API_KEY"]="openai api key"
os.environ["LANGCHAIN_API_KEY"]="langchain api key"
```

```python
from PyPDF2 import PdfReader

# Load PDF
pdf_path = "CAMasonFeb2024.pdf"
reader = PdfReader(pdf_path)

# Extract text from all pages
text = ""
for page in reader.pages:
    text += page.extract_text()
```

```python
text
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Create text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,    # number of characters in each chunk
    chunk_overlap=100   # overlap between chunks
)

# Split text
chunks = text_splitter.split_text(text)
```

```python
chunks
```

```python
from langchain_core.documents import Document

documents = [
    Document(page_content=chunk, metadata={"source": "pdf", "chunk_id": i})
    for i, chunk in enumerate(chunks)
]
```

```python
len(documents)
```

Vector store for embedding created by text-embedding-3-large

```python
from langchain_community.vectorstores import FAISS
from langchain_openai import OpenAIEmbeddings

# 1. Create embedding model
embedding_model = OpenAIEmbeddings(model="text-embedding-3-large")   # or any other embedding model

# 2. Build FAISS index from documents
faiss_index = FAISS.from_documents(documents, embedding_model)
```

```python
faiss_index.save_local("faiss_store1")
```

Vectore store for embedding created by mpnet model

```python
from langchain_community.embeddings import HuggingFaceEmbeddings


embedding_fn = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")

vectordb = FAISS.from_documents(documents, embedding_fn)
vectordb.save_local("mpnet")


```
